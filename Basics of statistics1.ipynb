Descriptive Analytics for Numerical Columns

```python
# Loading the dataset 
import pandas as pd
df=pd.read_csv("sales_data_with_discounts.csv")
df
```

```python
# Identify numerical columns in the dataset.
df.select_dtypes(include="number").columns
```

```python
# checking to fill missing values with mean or median
for x in df.select_dtypes(include="number").columns:
    print(f" the variable {x}:")
    print(f"the mean: {df[x].mean()}")
    print(f"the median: {df[x].median()}")
```

```python
# checking to fill missing values with mode 
for x in df.select_dtypes(include="number").columns:
    print(f"the mode:{df[x].mode()}")
```

```python
# checking to fill missing values with standard devivtion
for x in df.select_dtypes(include="number").columns:
     print(f"the standard devivtion:{df[x].std()}")
```

Provide a brief interpretation of these statistics

- The mean and median values helps to determine weather the data is in skewed or symmetric.

- If values are near to each other to fill null values mean is suitable.

- If outliers existed median values are suitable for missing values.

- Standard deviation measures how data distribution among the points from the center. it shows the deviation or spread of a data points.


Data Visualization

●	Histograms:


```python
# Histogram
import matplotlib.pyplot as plt
import numpy as np
for cols in df.select_dtypes(include="number").columns:
    plt.hist(df[cols])
    plt.title(f"hist of {cols}")
    plt.xlabel(cols)
    plt.ylabel("Frequency")
    plt.figure(figsize=(10,6))
    plt.show()
    
```

Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences.</h5>

- For continious variables by observing the histogram each and every variables are in positively skewed which means outliers are existed at the left side of the graph except for discount rate variable.

- For discount rate variable the histogram performs on negatively skewed which means ouliers are present in the right side of the graph.

- Due to this the analaysis can become difficult.

●	Box Plots


```python
# box plot
import matplotlib.pyplot as plt
import numpy as np
for cols in df.select_dtypes(include="number").columns:
    plt.boxplot(df[cols],vert=False)
    plt.title(f"boxplot of {cols}")
    plt.figure(figsize=(8,6))
    plt.show()
```

```python
import numpy as np
for cols in df.select_dtypes(include="number").columns:
    Q1 =np.percentile(df[cols],25)
    Q3 =np.percentile(df[cols],75)
    print("Q1:",Q1)
    print("Q3:",Q3)
    IQR = Q3-Q1
    UW= Q3+(1.5*IQR)
    print("Upper whisker:",UW)
    # outliers idenfication 
    print("Total number of outliers:",len(df[df[cols]>UW]))
```

Discuss any findings, such as extreme values or unusual distributions.

- I found that in each and every variables, the extreme values are found at the right side of the box plots except in discount rate variable.

- For discount rate variable the outliers are the extreme values are found at the left side of the box plot.

Bar Chart Analysis for Categorical Column

```python
# Seperating categorical variables from a dataset
x_disc=df[["Date", "Day", "SKU", "City", "BU", "Brand", "Model"]]
```

```python
# Bar chart 
for z in x_disc:
    x_disc[z].value_counts().plot(kind="bar",edgecolor="black")
    plt.xticks(rotation=45)
    plt.show()
```

Analyze the distribution of categories and provide insights.

- Some categories show higher frequency than others. so this categories are higly prefered by customers and less frequency does not impact the analysis.
