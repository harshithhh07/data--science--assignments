<h3>ESTIMATION AND CONFIDENCE INTERVALS</h3>

<h3>Data</h3>

<h4>A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29
</h4>

<h3>Assignment Tasks:</h3>

<h3>a. Build 99% Confidence Interval Using Sample Standard Deviation</h3>

<h4>Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.</h4>

```python
import pandas as pd

durability_list =[1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
durability_data = pd.Series(durability_list)
durability_data
```

```python
durability_data.info()
```

```python
durability_data.describe()
```

```python
import numpy as np
from scipy import stats

# Sample size
n = len(durability_data)

# Sample mean
sample_mean = np.mean(durability_data)

# Sample standard deviation
sample_std = np.std(durability_data, ddof=1) 

# Confidence level (99%)
confidence_level = 0.99
alpha = 1 - confidence_level

# Degrees of freedom
df = n - 1

# Critical t-value
t_critical = stats.t.ppf(1 - alpha/2, df)

# Standard error of the mean
standard_error = sample_std / np.sqrt(n)

# Margin of error
margin_of_error = t_critical * standard_error

# Confidence interval
conf_interval_lower = sample_mean - margin_of_error
conf_interval_upper = sample_mean + margin_of_error

print(f"Sample Size (n): {n}")
print(f"Sample Mean : {sample_mean:.2f} million characters")
print(f"Sample Standard Deviation (s): {sample_std:.2f} million characters")
print(f"Degrees of Freedom (df): {df}")
print(f"Critical t-value for 99% CI: {t_critical:.3f}")
print(f"Standard Error of the Mean: {standard_error:.3f}")
print(f"Margin of Error: {margin_of_error:.3f} million characters")
print(f"\n99% Confidence Interval: ({conf_interval_lower:.2f}, {conf_interval_upper:.2f}) million characters")

print(f"We are 99% confident that the true population mean durability of the print-heads is between {conf_interval_lower:.2f} and {conf_interval_upper:.2f} million characters.")
```

<h4>b. Build 99% Confidence Interval Using Known Population Standard Deviation
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.</h4>



```python
import numpy as np
from scipy import stats

# Known population standard deviation (in millions of characters)
population_std = 0.2

# Sample size (from previous analysis)
n = len(durability_data)

# Sample mean (from previous analysis)
sample_mean = np.mean(durability_data)

# Confidence level (99%)
confidence_level = 0.99
alpha = 1 - confidence_level

# Critical z-value for a 99% confidence interval
# stats.norm.ppf(1 - alpha/2) gives the z-value for the upper tail
z_critical = stats.norm.ppf(1 - alpha/2)

# Standard error of the mean (using population standard deviation)
standard_error_pop_std = population_std / np.sqrt(n)

# Margin of error
margin_of_error_pop_std = z_critical * standard_error_pop_std

# Confidence interval
conf_interval_lower_pop_std = sample_mean - margin_of_error_pop_std
conf_interval_upper_pop_std = sample_mean + margin_of_error_pop_std

print(f"Known Population Standard Deviation: {population_std:.2f} million characters")
print(f"Sample Size (n): {n}")
print(f"Sample Mean : {sample_mean:.2f} million characters")
print(f"Critical z-value for 99% CI: {z_critical:.3f}")
print(f"Standard Error of the Mean : {standard_error_pop_std:.3f}")
print(f"Margin of Error: {margin_of_error_pop_std:.3f} million characters")
print(f"\n99% Confidence Interval: ({conf_interval_lower_pop_std:.2f}, {conf_interval_upper_pop_std:.2f}) million characters")

print(f"We are 99% confident that the true population mean durability of the print-heads is between {conf_interval_lower_pop_std:.2f} and {conf_interval_upper_pop_std:.2f} million characters, assuming the population standard deviation is {population_std:.2f} million characters.")
```

<h3>HYPOTHESIS TESTING</h3>

<h4>Data Provided:</h4>

●	The theoretical weekly operating cost model: W = $1,000 + $5X

●	Sample of 25 restaurants with a mean weekly cost of Rs. 3,050

●	Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units


<h4>Assignment Tasks:</h4>

<h4>1. State the Hypotheses statement:</h4>

1. Null Hypothesis (H0): There is no significant difference in the average purchase value between the two groups.

    H0 :μ1 = μ2

2. Alternative Hypothesis (H1): There is a significant difference in the average purchase value between the groups.

    H1 : μ1 ≠ μ2

<h4>2. Calculate the Test Statistic:</h4>

Given:

X1 = 3050

𝜇 = 4000

σ= 5×25 = 125

n=25

```python
# Sample size 25 restaurants
n = 25

# Sample mean weekly cost Rs.3,050
X1 = 3050

# theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
𝜇 = 4000

# standard deviation (σ) of 25 units
sigma = 125

# Calculate the standard error of the mean
standard_error =  sigma / np.sqrt(n)

# Calculate the test statistic (z)
z_statistic = (X1 - 𝜇) / standard_error

print(f"Sample Size (n): {n}")
print(f"Sample Mean (X1): {X1} Rs")
print(f"Theoretical Mean (𝜇 ): {𝜇 } $")
print(f"Standard Deviation (sigma): {sigma} $")
print(f"Standard Error: {standard_error}")
print(f"Calculated Test Statistic (z): {z_statistic}")
```

<h4>3. Determine the Probability and compare:</h4>

Using the alpha level of 5% (α = 0.05),


```python
from scipy.stats import norm

alpha = 0.05 

# Calculate the p-value for a two-tailed test
p_value = 2 * norm.cdf(z_statistic)

print(f"Calculated Z-statistic: {z_statistic}")
print(f"Calculated P-value: {p_value:.10f}")
print(f"Significance Level (alpha): {alpha}")
print()

# Compare p-value with alpha
print("Hypothesis Test comparsion :")
if p_value < alpha:
    print("Since the P-value is less than alpha, we reject the null hypothesis.")
    print("This suggests a statistically significant difference between the sample mean and the theoretical mean.")
else:
    print("Since the P-value is greater than or equal to alpha, we fail to reject the null hypothesis.")
    print("This suggests no statistically significant difference between the sample mean and the theoretical mean.")

```

<h4>4. Make a Decision:
Compare the test statistic with the critical value to decide whether to reject the null hypothesis.</h4>


```python
from scipy.stats import norm

# Define alpha 
alpha = 0.05

# Calculate the critical z-values for a two-tailed test
# For a two-tailed test, each tail has alpha/2 probability
critical_z_lower = norm.ppf(alpha / 2)
critical_z_upper = norm.ppf(1 - (alpha / 2))

print(f"Calculated Z-statistic: {z_statistic}")
print(f"Significance Level (alpha): {alpha}")
print(f"Critical Z-values (two-tailed): {critical_z_lower:.3f} and {critical_z_upper:.3f}")
print()

# Compare z-statistic with critical values
print("Hypothesis Test Comparsion Critical Value :")
if z_statistic < critical_z_lower or z_statistic > critical_z_upper:
    print("Since the calculated Z-statistic falls outside the critical values, we reject the null hypothesis.")
   
else:
    print("Since the calculated Z-statistic falls within the critical values, we fail to reject the null hypothesis.")
```
