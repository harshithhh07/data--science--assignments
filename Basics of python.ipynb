Exercise 1: Prime Numbers

```python
# Prime numbers
a=int(input("enter any number:\n"))
if a > 1:
    for i in range(2, a):
        if a % i==0:
            print(f"{a} is not prime number")
            break
    else:
        print(f"{a} is a prime number")
else:
    print(f"{a} is not a prime number")
    
    
```

Exercise 2: Product of Random Numbers

```python
# Product of random numbers 
import random
a= random.randint(1,9)
b= random.randint(1,9)
user=int(input("enter your number:"))
if user == a * b:
    print("predicted correctly.")
else:
    print("predicted incorrectly.")      
```

Exercise 4:Hospital Billing System

```python
# Hospital billing program

# input parameters
x=int(input("enter the number of days stayed in hospital: ")) #number of days 
y=int(input("enter the number of lab test taken by a patient: ")) # number of tests
z=int(input("enter the medical charges : ")) #medicine charges
Room_charges_per_day=2000 *x 
Doctor_fee=1500
Lab_test_charges=300 *y
Medicine_charges= z

#calculating toatl bill
bill=Room_charges_per_day + Doctor_fee + Lab_test_charges + Medicine_charges

# checking the bill. if it is greater than 10k then applying 10% discount.
if bill>10000:
    print()
    print(" you have 10% of discount on charges ")
    discount=bill * (10/100)
else:
    discount=0

# calculating the final_bill
final_bill=bill - discount

# breakdown 
print("---- Hospital Bill Breakdown ----")
print()
print(f" Room charges ({x} days): {Room_charges_per_day}")
print(f" Doctor Fees : {Doctor_fee}")
print(f" lab test charges ({y} tests): {Lab_test_charges}")
print(f" medicine charges : {Medicine_charges}")
print(f" total bill before discount : {bill}")
print(f" discount applied : {discount}")
print(f" final bill : {final_bill}")

      

```
