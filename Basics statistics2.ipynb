```python
# append function
l1=[]
l1.append(10)
l1.append(20)
l1.append(30)
l1.append(40)
l1.append(50)
print(l1)
```

```python
# extend function
l1=["apple","banana","cherry"]
l2=["mango","orange","grapes"]
l1.extend(l2)
print(l1)
print(l2)
```

```python
# insert function
list1=[10,20,30,40]
print(list1)
list1.insert(1,15) # insert(index,element)
list1.insert(3,25)
list1.insert(5,35)
print(list1)
```

```python
# remove function
list1=[10,20,30,40,50,60]
print(list1)
list1.remove(30)
list1.remove(60)
print(list1)
```

```python
# pop function
l=[10,20,30,40,50]
print(l.pop())
print(l)
```

```python
# index function 
l=[10,20,30,40,50,60]
print(l.index(20))
```

```python
# count function 
l=[10,20,10,10,30,40,50,10,60]
print(l.count(10))
```

```python
# sort function 
l=[20,10,1,13,17,25,5,7]
l.sort()
print(l)
```

```python
# reverse function
l=[1,2,3,4,5,6]
l.reverse()
print(l)
```

```python
# clear function 
l=[10,20,30,40,50]
l.clear()
print(l)
```

Section C: Exploring Pandas

```python
# Create a small DataFrame manually with 10 rows and columns
data={
    "Name":["abhi","binnu","kanna","mahesh","cherry","siri","nannu","swapna","arjun","rohith"],
    "Age":[21,23,22,20,21,23,19,25,26,29],
    "City":["khammam","adilabad","nirmal","hyderabad","nizamabad","metpally","siddipet","kamareddy","nalgonda","nagarkurnool"],
    "Salary": [30000, 20000, None, 25000, None, 45000, 35000, 50000, None, 50000]
}
```

```python
# Loading data into dataframe 
import pandas as pd 
df =pd.DataFrame(data)
df
```

```python
# using .info to identifie the data types 
df.info()
```

```python
# .describe gives an basic stats representation of data 
df.describe()
```

```python
# selecting name and city columns 
df[["Name","City"]]
```

```python
# Dropping City column
df.drop("City",axis=1)
```

```python
# Filling the missing values in Salary column with the mean
df.fillna(df["Salary"].mean())
```

```python
# Removing duplicate rows
# there is no duplicate rows in this data 
# so we cannot remove duplicate rows
df.drop_duplicates()
```
