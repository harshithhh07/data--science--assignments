<h4>●	Discuss the pros and cons of One-Hot Encoding and Label Encoding.</h4>

<h5>One-Hot Encoding :</h5> 

1.Pros :

==> No ordinal relationship assumed.

==> Good for tree models.

==> Clear representation of each category.

2.Cons :

==> Increases number of columns (dummy variable explosion).

==> Not suitable for high-cardinality variables.

<h5>Label Encoding : </h5>
 
1.Pros :

==> Simple & fast.

==> No increase in number of features.

2.Cons :

==> Introduces artificial order (0 < 1 < 2).

==> May confuse linear models.

```python
# Observing the skeweness of age variable 
df['age'].skew()
```

```python
# Applying the log transformation for age variable 
import numpy as np 
df['age_log']= np.log1p(df['age'])
df
```

```python
# Showing the difference between age and age_log by histogram
Age=["age","age_log"]
for a in Age:
    if a == "Age":
        plt.hist(df[a])
        plt.title("histogram before applying the log transformation")
        plt.show()
    else:
        plt.hist(df[a])
        plt.title("histogram after applying the log transformation")
        plt.show()
```
