```python
# Estimation And Confidence Intervals
import numpy as np
from scipy import stats
Data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
# a. Build 99% Confidence Interval Using Sample Standard Deviation

mean = np.mean(Data)
print("mean is:",np.round(mean,2))
std = np.std(Data, ddof=1)   # sample standard deviation
print("sample standard deviation is:",np.round(std,2))
```

```python
n = len(Data) # Sample size
print(n)
df = n - 1
print(df)
```

```python
alpha = 0.01   # 99% confidence
t_critical = stats.t.ppf(1 - alpha/2, df) # t critical value
print(np.round(t_critical,2))
margin = t_critical * std / np.sqrt(n) # Margin of error
print(np.round(margin,2))
```

```python
# Confidence Interval
ci_lower = mean - margin
ci_upper = mean + margin
print(np.round(ci_lower,2))
print(np.round(ci_upper,2))
```

```python
# b. Build 99% Confidence Interval Using Known Population Standard Deviation

mean = np.mean(Data)
print("mean is:",mean)
sigma = 0.2  # known population standard deviation
n = len(Data)
print(n)
```

```python
z_critical = stats.norm.ppf(1 - alpha/2) # Z critical for 99% CI

margin = z_critical * sigma / np.sqrt(n) # margin of error

# confidence interval
ci_lower = mean - margin
ci_upper = mean + margin

(ci_lower, ci_upper)
```

```python
# Hypothesis Testing

import numpy as np
from scipy import stats
x_mean = 600 # sample mean
sample_mean = 3050 # sample mean weekly cost 
μ = 1000 + 5*x_mean # mean weekly cost 
σ = 5*25 # std dev of weekly cost
n = 25 # sample size
α = 0.05
```

```python
z_cal = (sample_mean - μ) / (σ / np.sqrt(n))
z_cal
```

```python
z_critical = stats.norm.ppf(1 - α)
z_critical
```

```python
p_value = 1 - stats.norm.cdf(z_cal)
p_value
```

```python
if p_value < α:
    print("Ho is rejected and H1 is accepted")
else:
    print("H1 is rejected and Ho is accepted")
```
